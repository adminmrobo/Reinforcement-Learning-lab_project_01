```python
import os
import subprocess
import platform
import sys
from tkinter import filedialog, Tk

def run_mrobo_hybrid_launcher_v8():
    # GUI 초기화
    root = Tk()
    root.withdraw()
    root.attributes("-topmost", True)
    
    print("\n" + "="*60)
    print(" 🤖 M-Robo [하이브리드 V11] 통합 런처 (V8-Hybrid)")
    print("="*60)

    # [1] 하이브리드 메인 스크립트 선택 (V11 우선 필터링)
    script_path = filedialog.askopenfilename(
        title="1. [하이브리드 V11] 메인 로봇 스크립트 선택",
        initialdir=os.getcwd(),
        filetypes=[
            ("하이브리드 메인", "hybrid_main_*.py"), 
            ("적극적 제어 메인", "fixed_main_robot_v8_*.py"), 
            ("파이썬 파일", "*.py")
        ]
    )
    if not script_path: root.destroy(); return

    # [2] 지능 모델 선택
    model_path = filedialog.askopenfilename(
        title="2. [교정 지능] 모델 선택 (fixed_master_...)",
        initialdir=os.path.dirname(script_path),
        filetypes=[
            ("교정 모델 파일", "fixed_master_intelligence_*.pkl"), 
            ("모든 모델", "*.pkl")
        ]
    )
    root.destroy()
    if not model_path: return

    # [3] 경로 및 실행 환경 정리
    script_dir = os.path.dirname(script_path)
    script_name = os.path.basename(script_path)
    abs_model_path = os.path.abspath(model_path)
    py_exec = sys.executable 

    os.chdir(script_dir)

    # [4] 독립 창 실행 (Windows 배치 파일 방식)
    if platform.system() == "Windows":
        # 타이틀에 HYBRID SYSTEM 명시
        bat_content = f"""@echo off
title M-Robo AI System [HYBRID MODE]
echo [System] M-Robo 하이브리드 시스템(V11) 가동 중...
echo [Mode] 단어 사전 우선 + AI 모델 보조
echo [Model] {os.path.basename(abs_model_path)}
echo --------------------------------------------------
"{py_exec}" "{script_name}" "{abs_model_path}"
pause
"""
        bat_path = os.path.join(script_dir, "mrobo_hybrid_run.bat")
        
        with open(bat_path, "w", encoding="cp949") as f:
            f.write(bat_content)

        # 새 창에서 로봇 메인 실행
        subprocess.Popen(f'start cmd /c "{bat_path}"', shell=True)
        
        # 뷰어/시뮬레이터 자동 가동 로직
        viewer_candidates = ['fixed_robot_viewer.py', 'new_robot_viewer.py', 'mrobo_simulator.py']
        for viewer in viewer_candidates:
            if os.path.exists(viewer):
                subprocess.Popen([py_exec, viewer])
                print(f"📺 뷰어 실행 중: {viewer}")
                break
                
        print(f"\n🚀 [하이브리드 V11] 기동 명령 전송 완료!")
        print(f"📂 작업 경로: {script_dir}")
        print(f"✨ 사전 매칭이 적용된 하이브리드 CMD 창을 확인해 주세요.")
        
    else:
        # Linux/Mac
        subprocess.Popen([py_exec, script_name, abs_model_path])

if __name__ == "__main__":
    run_mrobo_hybrid_launcher_v8()
```
